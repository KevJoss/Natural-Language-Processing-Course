## Part 2: Bag of Words Representation
---

Consider the following corpus of 3 documents

```python
Document_1 = "I love natural language processing"
Document_2 = "Natural language processing is fun"
Document_3 = "I love machine learning"
```

Perform tokenization and convert all words to lowercase

```python
from nltk.tokenize import word_tokenize

Document_1_tokens = word_tokenize(Document_1.lower())
Document_2_tokens = word_tokenize(Document_2.lower())
Document_3_tokens = word_tokenize(Document_3.lower())

print("Document 1 Tokens:", Document_1_tokens)
print("Document 2 Tokens:", Document_2_tokens)
print("Document 3 Tokens:", Document_3_tokens)
```

Build the vocabulary of the corpus

```python
corpus = [item for lst in [Document_1_tokens, Document_2_tokens, Document_3_tokens] for item in lst]
corpus = set(corpus)
print("Corpus:", corpus)
```

Construct the Bag-of-Words (BoW) representation as a term-document matrix

```python
from collections import defaultdict

def make_Bow(Corpus, Document):
    BoW = defaultdict(int)
    for element in Corpus:
        BoW[element]
        if element in Document:
            BoW[element] += 1
    return dict(BoW)
```

```python
BoW_Document_1 = make_Bow(corpus, Document_1_tokens)

print("BoW Document 1:", BoW_Document_1)

BoW_Document_2 = make_Bow(corpus, Document_2_tokens)

print("BoW Document 2:", BoW_Document_2)

BoW_Document_3 = make_Bow(corpus, Document_3_tokens)
print("BoW Document 3:", BoW_Document_3)
```

**What are the main limitations of the Bag-of-Words model?**
BoW have some limitations because of are a basic tecnhique to count the ocurrences of the words. Some limitations are the following:
1. BoWs trats a document as an unordered collection of words
2. It cannot undestand meaning, sarcasm, or relationships between words
3. Make BoWs, creeate large matrices that consume memory and computational resources
4. BoW ignores sentence structure, tense, negation and grammar
5. Small changes in wording create different features, such as "run", "running" and "ran"

## Part 3: Term Frequency (TF)
---

**Define Term Frequency (TF)**

Term Frequency is a measure used in text mining and information retrieval to indicate how often a word appears in a document. It helps determine the importance of a word within specific document.

The basic formula is:

$TF(t,d) = \frac{Number\ of\ times\ term\ t\ appears\ in\ document\ d}{Total\ number\ of\ terms\ in\ document\ d}$

Where:

$t = the\ term (word)$

$d = the\ document$

Compute the raw term frequency for each term in each document from Part 2

```python
def compute_TF(Document, BoW):
    return dict(BoW)
```

```python
TF_Document_1 = compute_TF(Document_1_tokens, BoW_Document_1)
print("TF Document 1:", TF_Document_1)

TF_Document_2 = compute_TF(Document_2_tokens, BoW_Document_2)
print("TF Document 2:", TF_Document_2)

TF_Document_3 = compute_TF(Document_3_tokens, BoW_Document_3)
print("TF Document 3:", TF_Document_3)
```

**Compute the normalized TF for each term using**:

$TF(t,d) = \frac{f_{t,d}}{\sum{t'}f_{t',d}}$

where $f_{t,d}$ is the frequency of term $t$ in document $d$

```python
def compute_TF_normalized(Document, BoW):
    TF_dict = defaultdict(int)
    number_total_terms = len(Document)
    
    for term, frequency in BoW.items():
        TF_dict[term] = frequency / number_total_terms
    return dict(TF_dict)
```

```python
TF_normalized_Document_1 = compute_TF_normalized(Document_1_tokens, BoW_Document_1)

print("TF Document 1:", TF_normalized_Document_1)

TF_normalized_Document_2 = compute_TF_normalized(Document_2_tokens, BoW_Document_2)
print("TF Document 2:", TF_normalized_Document_2)

TF_normalized_Document_3 = compute_TF_normalized(Document_3_tokens, BoW_Document_3)
print("TF Document 3:", TF_normalized_Document_3)
```

**Explain why normalization is useful.**

We normalize Term Frequency because raw word counts alone are not comparable across documents of different lenghts

## Part 4: Inverse Document Frequency (IDF)
---

**Define Inverse Document Frequency (IDF)**

Inverse Document Frequency (IDF) is a measure used in Information Retrieval and Natural Language Processing to determine how important or unique a word is across a collection of documents. The intuition behind IDF is inform how much important is a word that appears in many documents.

- High IDF -> The word is rare and important
- Low IDF -> The word is common and less useful for distinguishing documents

The $log$ is used mainly to smooth  the efect of rarity. 

Without the log, the ratio
$\frac{N}{df(t)}$

Where:
- $t$ = The term (word)
- $N$ = Total number of documents
- $df(t)$ = Number of document containing the term $t$

For example:
- A term appearing in 1 document out of 1,000,000 gets:
$\frac{1,000,000}{1} = 1,000,000$
- A term appearing 10 documents gets:
$\frac{1,000,000}{10} = 100,000$

Those raw values are huge and dominate the scoring too aggressively

Example with $N = 1,000,000$

|Document Frequency|Raw Ratio|Log IDF (base 10)|
|------------------|---------|-----------------|
|1|1,000,000|6|
|10|100,000|5|
|1,000|1,0000|3|
|100,000|10|1|

**Compute the IDF for each term using:**

$IDF(t) = \log(\frac{N}{df_{t}})$

where:
- $N$ is the total number of documents
- $df_{t}$ is the number of documents containing term t$t$

```python
Collection_Documents = [Document_1, Document_2, Document_3]
total_number_documents = len(Collection_Documents)

def df(term):
    count = 0
    for doc in Collection_Documents:
        if term in doc:
            count += 1
    return count
```

```python
import numpy as np

def IDF(term):
    return np.log(total_number_documents / df(term))
```

```python
for term in corpus:
    print(f"IDF of '{term}': {IDF(term)}")
```

**Which terms have the highest IDF? Why?**

The terms with highest IDF are: 
- *fun*
- *natural*
- *learning*
- *language*
- *is*
- *machine*

The main reason because are words with less frequency "1". Therefore have similarity importance for all words.

**What happens to the IDF of very common words?**

With very common words, the value of IDF decrease. These means that the term is a very common word, therefore have less importance respect documents collection.

## Part 5: TF-IDF Representation
---

**Define TF-IDF**

Term Frequency - Inverse Document Frequency is a statistical measure used in Information Retrieval and NLP to evaluate how important a word is to a document within a collection of documents (corpus). Combine Term Frequency (TF) and Inverse Document Frequency (IDF). TF-IDF is commonly used for:
- search engines
- document ranking
- keword extraction
- text classification
- recommendation systems
- document similarity

**Compute the $TF$-$IDF$ score for each term in each document:**

$TF$-$IDF(t,d) = TF(t,d) * IDF(t)$

```python
TF_IDF_Document_1 = {term: TF_normalized_Document_1[term] * IDF(term) for term in corpus}

print("TF-IDF Document 1:", TF_IDF_Document_1)

TF_IDF_Document_2 = {term: TF_normalized_Document_2[term] * IDF(term) for term in corpus}
print("TF-IDF Document 2:", TF_IDF_Document_2)

TF_IDF_Document_3 = {term: TF_normalized_Document_3[term] * IDF(term) for term in corpus}
print("TF-IDF Document 3:", TF_IDF_Document_3)
```

**Construct the TF-IDF matrix for the corpus.**

```python
TF_IDF_Document_1
```

```python
import pandas as pd

df_Document_1 = pd.DataFrame([TF_IDF_Document_1])
df_Document_2 = pd.DataFrame([TF_IDF_Document_2])
df_Document_3 = pd.DataFrame([TF_IDF_Document_3])

tfidf_matrix = pd.concat([df_Document_1, df_Document_2, df_Document_3])
tfidf_matrix.index = ['Document 1', 'Document 2', 'Document 3']

tfidf_matrix
```

The most important term in each document are the following:
- *Document 1*: "natural" is an important word for Document 1.
- *Document 2*: "natural, fun, is" are important words for Document 2.
- *Document 3*: "learning, machine" are important words for Document 3.

**Explain intuitively why TF-IDF works better than raw frequency.**

TF-IDF works better because consider statistical approaches unlike raw frequency that consider only the word's ocurrences. TF-IDF use two previous processes. "Term Frequency" and "Inverse Document Frequency" giving more details about the importance of a word considering the corpus and frequency.

**What are the main limitations of TF-IDF?**

TF-IDF ignores completely the semantics (synonyms, analogies, concepts, etc) of word because of only count words. For example when we mention car or automobile, the meaning refers to the same thing. However TF-IDF treats as different words. Also for words with the same syntaxis, but with different semantics depends on the contex (e.g. bank for river or financual institution).

Ignores word order and conext because of use BoW approach, and sentences with similiar representations that contain the same word can't be identified. For example
- "dog bites man" and "man bites dog"

If we have more words per document, the computational efficiency is affected in terms of memory. Finally TF-IDF prioritize rare words as very importants (e.g. typos, noise, very uncommon names), and not understand the importance of the document based on their authority, credibility and user intent.

**In what scenarios might TF-IDF fail?**

TF-IDF can fail or perform poorly in several real-world scenarios, especially when understanding meaning, context, or complex language patterns is important. For example when texts using different word for similar meanings. A lot of the scenarios about limitations of TF-IDF was detailed in the above question. However, some specific scenarios are specified below:

1. *Conversational AI and Chatbots*: TF-IDF cannot understand (intent, dialogue context, reasoning) and previuse conversational context.
2. *Recommendation Systems*: TF-IDF only analyzes text overlap, it cannot learn about user preferences, behavorial patterns and latent interests. For example two movies could be
very similar however TF-IDF can't identify because each movie use different vocabulary.
3. *Complex Question Answering*: TF-IDF cannot reason or combine facts to do inference about text or questions.

**Mention one modern alternative to TF-IDF (e.g., word embeddings) and briefly explain it.**

One modern alternative to TF-IDF is word embeddings, such as Word2Vec. This new approach represent words as dense numerical vectors in a continuous vector space. Unlinke TF-IDF,
embeddings capture semantic meaning and relationships between words. For example we can capture the following example:

- "king" and "queen"
- "car" and "automobile"

will have similar vector representations because they appear in similar contexts. Thanks to embeddings models can inference that words withm similar contexts tend to have similar meanings. Thanks to this approach embedding models can understand:
- synonyms
- semantic similarity
- analogies
- contextual relationships

Word embeddings are widely used in:
- semantic search
- recommendation systems
- chatbots
- machine translation
- modern NLP systems
